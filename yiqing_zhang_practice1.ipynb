```python
# Monte Carlo estimation of π
# Author: Yiqing Zhang

import numpy as np
import matplotlib.pyplot as plt
import math

N = 1_000_000   

x = np.random.rand(N)
y = np.random.rand(N)

count_inside = np.sum(x**2 + y**2 <= 1)

p_hat = count_inside / N
pi_est = 4 * p_hat

print(f"N = {N:,}")
print(f"Estimated π ≈ {pi_est:.6f}")

sample_size = 1000
inside_sample = x[:sample_size]**2 + y[:sample_size]**2 <= 1

plt.figure(figsize=(5,5))
plt.scatter(
    x[:sample_size], y[:sample_size],
    c=np.where(inside_sample, "tab:blue", "tab:red"),
    s=10, alpha=0.6, label='Random points'
)
circle = plt.Circle((0,0), 1, color='black', fill=False, lw=2, label='Quarter circle')
plt.gca().add_artist(circle)

plt.title(f"Monte Carlo π ≈ {pi_est:.5f}")
plt.xlabel("x")
plt.ylabel("y")
plt.axis("square")
plt.xlim(0,1)
plt.ylim(0,1)
plt.legend()
plt.show()
```
